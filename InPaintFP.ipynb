```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import lpips
import numpy as np
import cv2
import os
import matplotlib.pyplot as plt
from tqdm import tqdm
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
```

```python
class Places2Dataset(Dataset):
    def __init__(self, image_folder, transform=None):
        self.image_files = [os.path.join(image_folder, f) for f in os.listdir(image_folder) if f.endswith('.jpg')]
        self.transform = transform

    def __getitem__(self, idx):
        img = cv2.cvtColor(cv2.imread(self.image_files[idx]), cv2.COLOR_BGR2RGB)
        mask = np.ones_like(img) * 255
        cv2.rectangle(mask, (30, 30), (150, 150), (0, 0, 0), -1)
        if self.transform:
            img, mask = self.transform(img), self.transform(mask)
        return img * (mask > 0).float(), img, mask

    def __len__(self):
        return len(self.image_files)

transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((256, 256)),
    transforms.ToTensor(),
])

train_dataset = Places2Dataset('./dataset/val_256', transform=transform)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, num_workers=2, pin_memory=True)
```

```python
class UNetGenerator(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 64, 4, 2, 1), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(64, 128, 4, 2, 1), nn.BatchNorm2d(128), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(128, 256, 4, 2, 1), nn.BatchNorm2d(256), nn.LeakyReLU(0.2, inplace=True),
        )
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(256, 128, 4, 2, 1), nn.BatchNorm2d(128), nn.ReLU(inplace=True),
            nn.ConvTranspose2d(128, 64, 4, 2, 1), nn.BatchNorm2d(64), nn.ReLU(inplace=True),
            nn.ConvTranspose2d(64, 3, 4, 2, 1), nn.Tanh()
        )

    def forward(self, x):
        x = self.encoder(x)
        return self.decoder(x)


class PatchGANDiscriminator(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.Conv2d(3, 64, 4, 2, 1), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(64, 128, 4, 2, 1), nn.BatchNorm2d(128), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(128, 256, 4, 2, 1), nn.BatchNorm2d(256), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(256, 1, 4, 1, 1)  # No Sigmoid
        )

    def forward(self, x):
        return self.model(x)


# Initialize models
generator = UNetGenerator().cuda()
discriminator = PatchGANDiscriminator().cuda()
```

```python
# LPIPS Perceptual Loss
perceptual_loss_fn = lpips.LPIPS(net='vgg').cuda()

# Additional Losses - BCEWithLogitsLoss
adversarial_loss = nn.BCEWithLogitsLoss().cuda()
l1_loss = nn.L1Loss().cuda()
```

```python
# Optimizers for Generator and Discriminator
optimizer_G = optim.Adam(generator.parameters(), lr=0.0002, betas=(0.5, 0.999))
optimizer_D = optim.Adam(discriminator.parameters(), lr=0.0002, betas=(0.5, 0.999))
```

```python
import matplotlib.pyplot as plt
import os

# Create directories for saving checkpoints and results
os.makedirs("checkpoints", exist_ok=True)
os.makedirs("results", exist_ok=True)
```

```python
def save_checkpoint(epoch, generator, discriminator):
    #Save model checkpoints every 10 epochs
    torch.save(generator.state_dict(), f"checkpoints/generator_epoch_{epoch}.pth")
    torch.save(discriminator.state_dict(), f"checkpoints/discriminator_epoch_{epoch}.pth")
    print(f"Checkpoints saved for epoch {epoch}")
```

```python
def save_and_visualize_results(epoch, masked_img, fake_img, real_img):
    #Save and display inpainting results every 10 epochs
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    plt.title("Masked Image")
    plt.imshow(masked_img.permute(1,2,0).cpu().numpy())

    plt.subplot(1,3,2)
    plt.title("Generated Image (Inpainting)")
    plt.imshow(fake_img.permute(1,2,0).cpu().numpy())

    plt.subplot(1,3,3)
    plt.title("Original Image")
    plt.imshow(real_img.permute(1,2,0).cpu().numpy())

    # Save the image to results folder
    plt.savefig(f"results/epoch_{epoch}.png")
    plt.show()

    print(f"Results saved for epoch {epoch}")
```

```python
from tqdm import tqdm

# Number of epochs
epochs = 50

for epoch in range(1, epochs + 1):
    generator.train()
    discriminator.train()

    loop = tqdm(train_loader, leave=True)
    for masked_img, real_img, _ in loop:
        masked_img = masked_img.cuda()
        real_img = real_img.cuda()

        optimizer_D.zero_grad()
        real_validity = discriminator(real_img)
        fake_img = generator(masked_img)
        fake_validity = discriminator(fake_img.detach())

        d_loss = adversarial_loss(real_validity, torch.ones_like(real_validity)) + \
                 adversarial_loss(fake_validity, torch.zeros_like(fake_validity))
        d_loss.backward()
        optimizer_D.step()

        optimizer_G.zero_grad()
        fake_validity = discriminator(fake_img)
        g_loss = adversarial_loss(fake_validity, torch.ones_like(fake_validity)) + \
                 l1_loss(fake_img, real_img) * 100
        g_loss.backward()
        optimizer_G.step()

        loop.set_description(f"Epoch [{epoch}/{epochs}]")
        loop.set_postfix(d_loss=d_loss.item(), g_loss=g_loss.item())

    if epoch % 10 == 0:
        save_checkpoint(epoch, generator, discriminator)

    if epoch % 10 == 0:
        with torch.no_grad():
            fake_img = generator(masked_img)
            save_and_visualize_results(epoch, masked_img[0], fake_img[0], real_img[0])
```

```python

class Places2Dataset(Dataset):
    def __init__(self, image_folder, transform=None):
        self.image_files = sorted([os.path.join(image_folder, f)
                                   for f in os.listdir(image_folder)
                                   if f.endswith('.jpg')])[:10000]  # Use first 10,000 images only because less storage
        self.transform = transform

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img = cv2.cvtColor(cv2.imread(self.image_files[idx]), cv2.COLOR_BGR2RGB)
        mask = self.create_random_mask(img)

        if self.transform:
            img = self.transform(img)
            mask = self.transform(mask)

        masked_img = img * (mask > 0).float()
        return masked_img, img, mask

    def create_random_mask(self, img):
        #Create a random irregular mask
        mask = np.ones_like(img, dtype=np.uint8) * 255
        h, w = img.shape[:2]
        for _ in range(np.random.randint(1, 4)):
            x1, y1 = np.random.randint(0, w // 2), np.random.randint(0, h // 2)
            x2, y2 = np.random.randint(w // 2, w), np.random.randint(h // 2, h)
            cv2.rectangle(mask, (x1, y1), (x2, y2), (0, 0, 0), thickness=-1)
        return mask

#Image Transformations
transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((256, 256)),
    transforms.ToTensor(),
])

#Load Dataset of Places2Dataset (10,000 images)
dataset = Places2Dataset('./dataset/val_256', transform=transform)

#Split into train (80%) and test (20%)
train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size
train_dataset, test_dataset = random_split(dataset, [train_size, test_size])

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False, num_workers=2, pin_memory=True)

print(f"Dataset Loaded: {len(train_dataset)} train images, {len(test_dataset)} test images.")
```

```python
class MultiColumnEncoder(nn.Module):
    def __init__(self):
        super(MultiColumnEncoder, self).__init__()

        # Small Column
        self.small = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=5, stride=2, padding=2),
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=5, stride=2, padding=2),
            nn.ReLU()
        )

        # Medium Column
        self.medium = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=7, stride=2, padding=3),
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=7, stride=2, padding=3),
            nn.ReLU()
        )

        # Large Column
        self.large = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=9, stride=2, padding=4),
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=9, stride=2, padding=4),
            nn.ReLU()
        )

        # Merge Layer
        self.merge = nn.Sequential(
            nn.Conv2d(192, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.ReLU()
        )

    def forward(self, x):
        s = self.small(x)
        m = self.medium(x)
        l = self.large(x)
        merged = torch.cat([s, m, l], dim=1)  # Concatenate feature maps
        return self.merge(merged)
```

```python
class ResidualBlock(nn.Module):
    def __init__(self, in_channels):
        super(ResidualBlock, self).__init__()
        self.block = nn.Sequential(
            nn.Conv2d(in_channels, in_channels, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(in_channels, in_channels, kernel_size=3, padding=1)
        )

    def forward(self, x):
        return x + self.block(x)

class GMCNNGenerator(nn.Module):
    def __init__(self):
        super(GMCNNGenerator, self).__init__()
        self.encoder = MultiColumnEncoder()

        self.res_blocks = nn.Sequential(
            ResidualBlock(128),
            ResidualBlock(128),
            ResidualBlock(128)
        )

        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(128, 64, kernel_size=4, stride=2, padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(64, 3, kernel_size=4, stride=2, padding=1),
            nn.Tanh()
        )

    def forward(self, x):
        enc = self.encoder(x)
        res = self.res_blocks(enc)
        return self.decoder(res)
```

```python
class PatchGANDiscriminator(nn.Module):
    def __init__(self):
        super(PatchGANDiscriminator, self).__init__()
        self.model = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=4, stride=2, padding=1),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(64, 128, kernel_size=4, stride=2, padding=1),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(128, 256, kernel_size=4, stride=2, padding=1),
            nn.BatchNorm2d(256),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(256, 512, kernel_size=4, stride=1, padding=1),
            nn.BatchNorm2d(512),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(512, 1, kernel_size=4, stride=1, padding=1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.model(x)
```

```python
# Initialize GMCNN Generator and PatchGAN Discriminator
gmcnn_generator = GMCNNGenerator().cuda()
patchgan_discriminator = PatchGANDiscriminator().cuda()

print("GMCNN Generator and PatchGAN Discriminator initialized.")
```

```python
import torch.optim as optim
```

```python
# L1 Loss for pixel reconstruction
loss_l1 = nn.L1Loss().cuda()

# Adversarial Loss for GAN
loss_adv = nn.BCEWithLogitsLoss().cuda()

# LPIPS for Perceptual Loss
loss_perceptual = lpips.LPIPS(net='vgg').cuda()
```

```python
# Optimizers for GMCNN Generator and Discriminator
gmc_optimizer_G = optim.Adam(gmcnn_generator.parameters(), lr=0.0002, betas=(0.5, 0.999))
gmc_optimizer_D = optim.Adam(patchgan_discriminator.parameters(), lr=0.0002, betas=(0.5, 0.999))

# Learning Rate Scheduler for GMCNN
gmc_scheduler_G = torch.optim.lr_scheduler.StepLR(gmc_optimizer_G, step_size=20, gamma=0.5)
gmc_scheduler_D = torch.optim.lr_scheduler.StepLR(gmc_optimizer_D, step_size=20, gamma=0.5)

# Store Losses for Visualization
gmc_l1_losses, gmc_gan_losses, gmc_perceptual_losses = [], [], []

gmc_epochs = 100  # Train for 100 epochs

for epoch in range(1, gmc_epochs + 1):
    gmcnn_generator.train()
    patchgan_discriminator.train()

    epoch_gmc_l1_loss = 0
    epoch_gmc_gan_loss = 0
    epoch_gmc_perceptual_loss = 0

    loop = tqdm(train_loader, leave=True)

    for masked_img, real_img, _ in loop:
        masked_img, real_img = masked_img.cuda(), real_img.cuda()

        # Train GMCNN Discriminator
        gmc_optimizer_D.zero_grad()
        real_validity = patchgan_discriminator(real_img)
        fake_img = gmcnn_generator(masked_img)
        fake_validity = patchgan_discriminator(fake_img.detach())
        real_loss = loss_adv(real_validity, torch.ones_like(real_validity))
        fake_loss = loss_adv(fake_validity, torch.zeros_like(fake_validity))
        gmc_d_loss = (real_loss + fake_loss) / 2
        gmc_d_loss.backward()
        gmc_optimizer_D.step()

        # Train GMCNN Generator
        gmc_optimizer_G.zero_grad()
        fake_validity = patchgan_discriminator(fake_img)
        gmc_l1_loss_value = loss_l1(fake_img, real_img)
        gmc_perceptual_loss_value = loss_perceptual(fake_img, real_img).mean()
        gmc_g_loss = loss_adv(fake_validity, torch.ones_like(fake_validity)) + gmc_l1_loss_value * 50 + gmc_perceptual_loss_value * 20
        gmc_g_loss.backward()
        gmc_optimizer_G.step()

        epoch_gmc_l1_loss += gmc_l1_loss_value.item()
        epoch_gmc_gan_loss += gmc_d_loss.item()
        epoch_gmc_perceptual_loss += gmc_perceptual_loss_value.item()

    gmc_scheduler_G.step()
    gmc_scheduler_D.step()
    gmc_l1_losses.append(epoch_gmc_l1_loss / len(train_loader))
    gmc_gan_losses.append(epoch_gmc_gan_loss / len(train_loader))
    gmc_perceptual_losses.append(epoch_gmc_perceptual_loss / len(train_loader))

    if epoch % 10 == 0:
        torch.save(gmcnn_generator.state_dict(), f"checkpoints/gmc_generator_epoch_{epoch}.pth")
        torch.save(patchgan_discriminator.state_dict(), f"checkpoints/gmc_discriminator_epoch_{epoch}.pth")
        print(f"GMCNN Checkpoints saved for epoch {epoch}")

        with torch.no_grad():
            result_img = gmcnn_generator(masked_img)
            plt.figure(figsize=(12, 4))
            plt.subplot(1, 3, 1)
            plt.title("Masked Image")
            plt.imshow(masked_img[0].permute(1, 2, 0).cpu().numpy())
            plt.subplot(1, 3, 2)
            plt.title("Generated Image")
            plt.imshow(result_img[0].permute(1, 2, 0).cpu().numpy())
            plt.subplot(1, 3, 3)
            plt.title("Original Image")
            plt.imshow(real_img[0].permute(1, 2, 0).cpu().numpy())
            plt.savefig(f"results/gmc_epoch_{epoch}.png")
            plt.show()

        plt.figure(figsize=(10, 5))
        plt.plot(gmc_l1_losses, label='GMCNN L1 Loss')
        plt.plot(gmc_gan_losses, label='GMCNN GAN Loss')
        plt.plot(gmc_perceptual_losses, label='GMCNN Perceptual Loss')
        plt.xlabel('Epochs')
        plt.ylabel('Loss')
        plt.title('GMCNN Loss Curves')
        plt.legend()
        plt.grid(True)
        plt.show()

print("GMCNN Training Completed with Checkpoints and Results saved every 10 epochs.")
```

```python
import cv2
import torch
import matplotlib.pyplot as plt
import torch.nn.functional as F
from torchvision import transforms
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import numpy as np

# Load trained GMCNN model
gmcnn_generator = GMCNNGenerator().cuda()
gmcnn_generator.load_state_dict(torch.load('checkpoints/gmc_generator_epoch_100.pth'))
gmcnn_generator.eval()

# Image transformation
transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((256, 256)),
    transforms.ToTensor()
])

# Load custom image
img_path = 'testingdata2.png'
original = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
masked = original.copy()

# Apply mask
h, w = masked.shape[:2]
x1, y1 = w // 4, h // 4
x2, y2 = w // 2, h // 2
masked[y1:y2, x1:x2] = 0

# Transform images
original_tensor = transform(original).unsqueeze(0).cuda()
masked_tensor = transform(masked).unsqueeze(0).cuda()
masked_tensor.requires_grad = True

# Clean Inpainting
with torch.no_grad():
    clean_output = gmcnn_generator(masked_tensor)

# FGSM ATTACK
output_fgsm = gmcnn_generator(masked_tensor)
loss_fgsm = F.mse_loss(output_fgsm, original_tensor)
loss_fgsm.backward()

epsilon = 0.03
fgsm_input = masked_tensor + epsilon * masked_tensor.grad.sign()
fgsm_input = torch.clamp(fgsm_input, 0, 1).detach()

# PGD ATTACK
def pgd_attack(model, original_input, target, eps=0.03, alpha=0.01, iters=10):
    adv = original_input.clone().detach().requires_grad_(True)
    for _ in range(iters):
        output = model(adv)
        loss = F.mse_loss(output, target)
        loss.backward()
        adv_data = adv + alpha * adv.grad.sign()
        eta = torch.clamp(adv_data - original_input, -eps, eps)
        adv = torch.clamp(original_input + eta, 0, 1).detach().requires_grad_(True)
    return adv.detach()

pgd_input = pgd_attack(gmcnn_generator, masked_tensor, original_tensor)

# Inpainting with FGSM and PGD inputs
with torch.no_grad():
    fgsm_output = gmcnn_generator(fgsm_input)
    pgd_output = gmcnn_generator(pgd_input)

# Evaluation Metrics
def get_metrics(pred, target):
    pred_np = pred.squeeze(0).permute(1, 2, 0).cpu().numpy()
    target_np = target.squeeze(0).permute(1, 2, 0).cpu().numpy()

    return {
        "PSNR": psnr(target_np, pred_np, data_range=1.0),
        "SSIM": ssim(target_np, pred_np, data_range=1.0, channel_axis=-1)
    }


metrics_clean = get_metrics(clean_output, original_tensor)
metrics_fgsm = get_metrics(fgsm_output, original_tensor)
metrics_pgd = get_metrics(pgd_output, original_tensor)

# Print Metrics
print(" PSNR & SSIM Comparison:")
print(f"Clean  -> PSNR: {metrics_clean['PSNR']:.2f}, SSIM: {metrics_clean['SSIM']:.4f}")
print(f"FGSM   -> PSNR: {metrics_fgsm['PSNR']:.2f}, SSIM: {metrics_fgsm['SSIM']:.4f}")
print(f"PGD    -> PSNR: {metrics_pgd['PSNR']:.2f}, SSIM: {metrics_pgd['SSIM']:.4f}")

# Visualization
def show_image(tensor, title, index):
    plt.subplot(1, 5, index)
    plt.title(title)
    img = tensor.squeeze(0).permute(1, 2, 0).detach().cpu().numpy()
    plt.imshow(img)
    plt.axis('off')

plt.figure(figsize=(20, 4))
show_image(original_tensor, "Original", 1)
show_image(masked_tensor, "Masked", 2)
show_image(clean_output, "Clean Inpaint", 3)
show_image(fgsm_output, "FGSM Inpaint", 4)
show_image(pgd_output, "PGD Inpaint", 5)
plt.tight_layout()
plt.show()
```
